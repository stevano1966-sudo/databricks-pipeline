```sql
-- ============================================================
-- Script : 03_copy_to_staging
-- Purpose: Full overwrite copy from brewflow_oltp to brewflow_staging
-- Depends: brewflow_oltp (customers, stores, products, orders, order_items)
-- Outputs: brewflow_staging (stg_customers, stg_stores, stg_products,
--          stg_orders, stg_order_items)
-- ============================================================

INSERT OVERWRITE brewflow_staging.stg_customers
    SELECT * FROM brewflow_oltp.customers;

INSERT OVERWRITE brewflow_staging.stg_stores
    SELECT * FROM brewflow_oltp.stores;

INSERT OVERWRITE brewflow_staging.stg_products
    SELECT * FROM brewflow_oltp.products;

INSERT OVERWRITE brewflow_staging.stg_orders
    SELECT * FROM brewflow_oltp.orders;

INSERT OVERWRITE brewflow_staging.stg_order_items
    SELECT * FROM brewflow_oltp.order_items;

-- Validieren
SELECT 'stg_customers'   AS tbl, COUNT(*) AS rows FROM brewflow_staging.stg_customers
UNION ALL SELECT 'stg_stores',      COUNT(*) FROM brewflow_staging.stg_stores
UNION ALL SELECT 'stg_products',    COUNT(*) FROM brewflow_staging.stg_products
UNION ALL SELECT 'stg_orders',      COUNT(*) FROM brewflow_staging.stg_orders
UNION ALL SELECT 'stg_order_items', COUNT(*) FROM brewflow_staging.stg_order_items;
-- Erwartet: 10, 5, 8, 22, 32
```
