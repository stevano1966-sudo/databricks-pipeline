```sql
-- ============================================================
-- Script : 03_copy_to_staging_northriver
-- Purpose: Full overwrite copy from northriver_oltp to northriver_staging
-- Depends: northriver_oltp (stg_customers, stg_products, stg_orders, stg_order_items)
-- Outputs: northriver_staging (stg_customers, stg_products, stg_orders, stg_order_items)
-- Hinweis: kein stores/stg_stores/dim_store in diesem Projekt (weder OLTP
--          noch Staging noch Warehouse) -> Zeilen dafür entfernt.
-- ============================================================

INSERT OVERWRITE workspace.northriver_staging.stg_customers
    SELECT * FROM workspace.northriver_oltp.stg_customers;

INSERT OVERWRITE workspace.northriver_staging.stg_products
    SELECT * FROM workspace.northriver_oltp.stg_products;

INSERT OVERWRITE workspace.northriver_staging.stg_orders
    SELECT * FROM workspace.northriver_oltp.stg_orders;

INSERT OVERWRITE workspace.northriver_staging.stg_order_items
    SELECT * FROM workspace.northriver_oltp.stg_order_items;

-- Validieren
SELECT 'stg_customers'   AS tbl, COUNT(*) AS rows FROM workspace.northriver_staging.stg_customers
UNION ALL SELECT 'stg_products',    COUNT(*) FROM workspace.northriver_staging.stg_products
UNION ALL SELECT 'stg_orders',      COUNT(*) FROM workspace.northriver_staging.stg_orders
UNION ALL SELECT 'stg_order_items', COUNT(*) FROM workspace.northriver_staging.stg_order_items;
-- Erwartet: 18, 14, 29, 32
```
